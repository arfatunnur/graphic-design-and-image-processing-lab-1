```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('/content/apple.jpg')
image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
h,w=image.shape[:2]
```

```python
sx=2
sy=1.5
```

```python
scaled = cv2.resize(image,None, fx=sx, fy=sy)
```

```python
h2,w2 = scaled.shape[:2]
```

```python
center = (w2//2,h2//2)
```

```python
tx=50
ty=30
```

```python
M_rotation = np.float32([
    [1,0,tx],
    [0,1,ty]
])
```

```python
rotated = cv2.warpAffine(scaled, M_rotation, (w2,h2))
```

```python
M_translation = np.float32([
    [1,0,tx],
    [0,1,ty]
])
```

```python
translated = cv2.warpAffine(
    image,
    M_rotation,
     (w2+tx,h2+ty))
```

```python
reflected = cv2.flip(image, 0)
```

```python
plt.figure(figsize=(15,5))
```

```python
plt.subplot(2,3,1)
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')

plt.subplot(2,3,2)
plt.imshow(translated)
plt.title('Translated')
plt.axis('off')

plt.subplot(2,3,3)
plt.imshow(rotated)
plt.title('Rotated')
plt.axis('off')

plt.subplot(2,3,4)
plt.imshow(reflected)
plt.title('Reflection about X-axis')
plt.axis('off')

plt.subplot(2,3,5)
plt.imshow(scaled)
plt.title('Scaled')
plt.axis('off')
```
